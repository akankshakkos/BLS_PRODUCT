```sql
%python
import urllib.request

BLS_URL = (
    "https://download.bls.gov/pub/time.series/pr/"
)

print(BLS_URL)
```

```sql
%python
try:

    response = urllib.request.urlopen(
        BLS_URL,
        timeout=30
    )

    print("HTTP status:", response.status)

    html = response.read().decode(
        "utf-8",
        errors="replace"
    )

    print(html[:1000])

except Exception as e:

    print("ERROR:")
    print(type(e).__name__)
    print(str(e))
```
